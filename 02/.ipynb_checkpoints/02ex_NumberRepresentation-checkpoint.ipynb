1\. Write a function that **converts number representation**, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
# 4-bit binary → hex lookup
bin_hex = {format(i, '04b'): format(i, 'X') for i in range(16)}

def conv_rep(x, in_fmt, out_fmt):
    
    # DECIMAL CONVERSION
    if in_fmt == 'd':
        # Convert decimal to binary string
        binary = []
        while x != 0:
            binary.append(str(x % 2))
            x //= 2
        binary.reverse()
        bin_str = ''.join(binary)

        if out_fmt == 'bin':
            return bin_str

        elif out_fmt == 'hex':
            # Pad binary string to multiple of 4
            while len(bin_str) % 4 != 0:
                bin_str = '0' + bin_str
            # Convert each 4-bit block to hex
            hexa = ''.join(bin_hex[bin_str[i:i+4]] for i in range(0, len(bin_str), 4))
            return hexa

    # BINARY CONVERSION
    if in_fmt == 'bin':
        bin_num = x[2:]
        d = 0
        for i in range(len(bin_num)):
            d += int(bin_num[i], 2) * pow(2, len(bin_num) - i - 1)

        if out_fmt == 'd':
            return d
        elif out_fmt == 'hex':
            return conv_rep(d, 'd', 'hex')

    # HEXA CONVERSION
    if in_fmt == 'hex':
        hexa_num = x[2:]
        d = 0
        for i in range(len(hexa_num)):
            d += int(hexa_num[i], 16) * pow(16, len(hexa_num) - i - 1)

        if out_fmt == 'd':
            return d
        elif out_fmt == 'bin':
            return conv_rep(d, 'd', 'bin')
```

```python
print(conv_rep(10, 'd', 'bin'),   # '1010'
        conv_rep(10, 'd', 'hex'),   # 'A'
        conv_rep(255, 'd', 'bin'),  # '11111111'
        conv_rep(255, 'd', 'hex'))  # 'FF'
```

2\. Write a function that converts a **32 bit word** into a single precision **floating point** (i.e. interprets the various bits as sign, mantissa and exponent)

```python
def floating(word): 

    sign = int(word[0])
    exponent = int(word[1:9],2)
    mantissa = word[9:]

    mant_real = 1 #-> starts from 1
    for i, val in enumerate(mantissa): 
        mant_real += int(val)*(2**(-i - 1))
    n = (-1)**sign * mant_real *2 **(exponent - 127)
    return n

print(floating('01000000101110000000000000000000')) # 5.75
```

5\. Write a function that takes in input three parameters $a$, $b$ and $c$ and prints out the two **solutions to the quadratic** equation $ax^2+bx+c=0$ using the standard formula:
$$
x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$

(a) use the program to compute the solution for $a=0.001$, $b=1000$ and $c=0.001$

```python
import numpy as np

def quad_sol(a, b, c):
    xp = (-b + np.sqrt(b * b - 4 * a * c)) / (2 * a)
    xm = (-b - np.sqrt(b * b - 4 * a * c)) / (2 * a)
    print(f"Solutions for eq: {a}x^2 + {b}x + {c} = 0")
    return xp, xm
```

```python
quad_sol(0.001, 1000, 0.001)
```

(b) re-express the standard solution formula by multiplying top and bottom by $-b\mp\sqrt{b^2-4ac}$ and again find the solution for $a=0.001$, $b=1000$ and $c=0.001$. How does it compare with what previously obtained? Why?

```python
def quad_sol2(a, b, c): # numerical stable version
    yp, ym = -b - np.sqrt(b * b - 4 * a * c), -b + np.sqrt(b * b - 4 * a * c)
    xp = 2 * c / yp
    xm = 2 * c / ym
    
    print(f"Solutions for eq: {a}x^2 + {b}x + {c} = 0")
    return xp, xm
```

```python
quad_sol2(0.001, 1000, 0.001)

# We want to avoid the subtraction of two similar very large numbers, both at num or den. -> Catastophic cancellation
# So even if the first algo is algebraic identically to the second, we have:
# First algoritm stable for the larger root (x2) because we add two identical number, but is unstable for x1.
# The second one inverts the problem: it is more stable for the smaller root (x1) because it is obtained dividing by the addition of two identical numbers.
```

(c) write a function that compute the **roots of a quadratic equation accurately in all cases**

```python
def quad_sol_def(a, b, c):
    y = (-b - np.sqrt(b * b - 4 * a * c)) # stable calculation
    xm = y / (2 * a) # from 1
    xp = 2 * c / y # from 2

    return xp, xm
```

```python
quad_sol_def(0.001, 1000, 0.001)
```

6\. Write a program that implements the function **$f(x)=x(x−1)$**

(a) Calculate the derivative of the function at the point $x = 1$ using the derivative definition:

$$
\frac{{\rm d}f}{{\rm d}x} = \lim_{\delta\to0} \frac{f(x+\delta)-f(x)}{\delta}
$$

with $\delta = 10^{−2}$. Calculate the true value of the same derivative analytically and compare with the answer your program gives. The two will not agree perfectly. Why not?

```python
f = lambda x : x * (x-1)
```

```python
der = lambda x, delta: (f(x + delta) - f(x)) / delta

print(f"f(1) = {f(1)}")
print(f"Analitical derivative at x=1: {1}")
print(f"Limit derivative at x=1: {der(1, 0.01)}")
```

(b) Repeat the calculation for $\delta = 10^{−4}, 10^{−6}, 10^{−8}, 10^{−10}, 10^{−12}$ and $10^{−14}$. How does the accuracy scales with $\delta$?

```python
import matplotlib.pyplot as plt

delta = [pow(10, -i) for i in range(2, 16, 2)]

plt.figure(figsize = (10, 6))
plt.suptitle("Derivative limit f(x) = x(x-1) at x=1")
plt.xlabel("$\delta$")
plt.ylabel("df/dx(x; $\delta$)")
plt.xscale('log')
plt.axhline(1)

for d in delta:
    plt.scatter(d, der(1, d), marker='x', color='red')

plt.tight_layout()
plt.show()
```

7\. Consider the integral of the **semicircle of radius 1**:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

```python
f = lambda x: np.sqrt(1 - x*x)
```

```python
def integral(N):
    h = 2 / N
    I = 0
    for i in range(1, N+1):
        I += h * f(-1 + i*h)
    return I
```

```python
integral(200)
```

b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute?

```python
import time

N = 100
t0 = time.time()
I = integral(N)
T = time.time() - t0

while T < 1:
    N_prec, I_prec, T_prec = N, I, T
    N += 10
    t0 = time.time()
    I = integral(N)
    T = time.time() - t0

print(f"N max under 1 second: = {N_prec}")
print(f"Integral value = {I_prec}")
print(f"Computational time = {T_prec}")
```

```python
N = 100
t0 = time.time()
I = integral(N)
T = time.time() - t0

while T < 60:
    N_prec_60, I_prec_60, T_prec_60 = N, I, T
    N *= 10
    t0 = time.time()
    I = integral(N)
    T = time.time() - t0

print(f"N max under 1 second: = {N_prec_60}")
print(f"Integral value = {I_prec_60}")
print(f"Computational time = {T_prec_60}")
```
