1\. Find the **row, column and overall means** for the following matrix:

```python
m = np.arange(12).reshape((3,4))
```

````python
for i in range(m.shape[0]):
    print(f"Mean of row {i} = {m[i,:].mean()}")
    
for j in range(m.shape[1]):
    print(f"Mean of column {j} = {m[:,j].mean()}")

print(f"Overall mean = {m.mean()}")
````

2\. Find the **outer product** of the following two vecotrs:

````python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
````

````python
#Using the function outer in numpy

out = np.outer(u,v)
print(out)
````

````python
#Using a nested for loop or list comprehension
out = np.zeros((u.shape[0],v.shape[0]))
for i in range(out.shape[0]):
    for j in range(out.shape[1]):
        out[i,j] = u[i]*v[j]

print(out)
````

````python
#Using list comprehension

out = [[int(u[i]*v[j]) for j in range(v.shape[0])] for i in range(u.shape[0])]
print(np.array(out).reshape(4,4))
````

````python
#Using numpy broadcasting operatoins
u_broad = np.tile(u, (v.shape[0], 1)).T # we want u in 4 concatenated lines, then transpose it
out = u_broad*v
print(out)
````

4\. Use np.linspace to create an array of **100 numbers between 0 and 2π** (includsive).

  * Extract every 10th element using slice notation
  * Reverse the array using slice notation
  * Extract elements where the absolute difference between the sine and cosine functions evaluated at that element is less than 0.1
  * Make a plot showing the sin and cos functions and indicate where they are close

````python
n = np.linspace(0, 2*pi, 100)
print(f"Extract every 10th element using slice notation: \n {n[::10]} \n")
print(f"Reverse the array using slice notation: \n {n[::-1]}")
````

````python
mask = (np.abs( np.sin(n) - np.cos(n) ) < 0.1)
print(f"Values for which |sin - cos| < 0.1: \n {n[mask]}")
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.scatter(n, np.sin(n), color='r', label='sin(x)')
plt.scatter(n, np.cos(n), color='g', label='cos(x)')
for x in n[mask]:
    plt.axvline(x, ymin=0, ymax=1, color='b')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.tight_layout
plt.show()
````

5\. Create a matrix that shows the **10 by 10 multiplication table**.

 * Find the trace of the matrix
 * Extract the anto-diagonal (this should be ```array([10, 18, 24, 28, 30, 30, 28, 24, 18, 10])```)
 * Extract the diagnoal offset by 1 upwards (this should be ```array([ 2,  6, 12, 20, 30, 42, 56, 72, 90])```)

````python
m = np.outer(np.arange(1, 11), np.arange(1, 11))
print(f"10 by 10 multiplication table: \n {m} \n")

ad = np.array([m[i, 9 - i] for i in range(10)])
print(f"antodiagonal: \n {ad} \n")

d_off = np.array([m[i + 1, i] for i in range(9)])
print(f"offset diagonal: \n {d_off} \n")
````

6\. Use broadcasting to create a **grid of distances**

Route 66 crosses the following cities in the US: Chicago, Springfield, Saint-Louis, Tulsa, Oklahoma City, Amarillo, Santa Fe, Albuquerque, Flagstaff, Los Angeles
The corresponding positions in miles are: 0, 198, 303, 736, 871, 1175, 1475, 1544, 1913, 2448

  * Construct a 2D grid of distances among each city along Route 66
  * Convert that in km (those savages...)

````python
d = np.array([0, 198, 303, 736, 871, 1175, 1475, 1544, 1913, 2448])
cities = ["Chicago", "Springfield", "Saint-Louis", "Tulsa", "Oklahoma City", "Amarillo", "Santa Fe", "Albuquerque", "Flagstaff", "Los Angeles"]
cities_dict = {city: i for i, city in enumerate(cities)}
````

````python
d_tile = np.tile(d, (d.shape[0], 1)).T # the same as exercise 2
d_grid = np.abs(d_tile - d)
print(f"2D distance Grid: \n {d_grid} \n") 

conv = 1.61 # km / miles
print(f"2D distance Grid in km: \n {d_grid * conv} \n") # in km
````

````python
def distance(city1, city2):
    d = float(d_grid[cities_dict[city1], cities_dict[city2]] * conv)
    print(f"Distance between {city1} and {city2}:= {d:.2f} km")

distance("Chicago", "Amarillo")
````

7\. Prime numbers sieve: compute the prime numbers in the 0-N (N=99 to start with) range with a sieve (mask).
  * Constract a shape (100,) boolean array, the mask
  * Identify the multiples of each number starting from 2 and set accordingly the corresponding mask element
  * Apply the mask to obtain an array of ordered prime numbers
  * Check the performances (timeit); how does it scale with N?
  * Implement the optimization suggested in the [sieve of Eratosthenes](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes)

````python
# With a boolean mask
N_list = [10**i for i in range(2,5)]

for N in N_list:
    n = np.arange(0, N)
    mask = np.ones(N, 'bool')
    
    for i in range(2, N):
        for j in range(2, N):
            if ((n[j] % i == 0) and (n[j] != i)):
                mask[j] = False
    print(f"Prime numbers in 0-{N}: \n {n[mask]} \n with computational time:")
    %timeit n[mask]
````

````python
# With the Sieve of Eratosthenes

def sieve(N):
    n_prime = np.arange(0, N).tolist()
    for n in n_prime:
        for i in range(2, int(N / 2)): #excluding 0, 1
            if not (n*i in n_prime):
                continue
            else:
                n_prime.remove(n*i)
%timeit sieve(99)
````
