2\. Create a random matrix of 5x5 and then save it to a text file named **"data.txt"**

```bash
touch data.txt
```

````python
m = np.random.rand(5,5)

with open('data.txt', mode='w') as f: # write
    for i in range(m.shape[0]):
        for j in range(m.shape[1]):
            f.write(f"{m[i,j]} ")
        f.write("\n")
````

````python
with open('data.txt', mode='r') as f: # read
    for line in f:
        print(line)
````

3\. Load the saved txt file of point 2 and **convert it to a csv file** (by hand)

```bash
touch data.csv
```

````python
lines = []
with open('data.txt', mode='r') as txt:
    for line in txt:
        lines.append(line)

print(lines)

# now we have to format txt lines as csv lines
with open('data.csv', mode='w') as csv:
    for line_str in lines:
            line_str = line_str.replace(' ', ',') # change delimeter returning the new one
            csv.write(line_str)
````

````python
with open('data.csv', mode='r') as csv:
    for line in csv:
        print(line)
````

4\. load the **binary file** named *credit_card.dat* and convert the data into the real credit-card number.
Each line correspond to a credit card number.
Each character is composed by 6 bit (even the space) and the last 4 bit are just a padding

**hint**: use the `chr()` function to convert a number to a char

````python
line_list = []
with open('credit_card.dat', mode='rb') as f: #read it as strings
    for line in f:
        line_list.append(line)
        
line_list = line_list[0:-1] #remove the last line (onhly 4 digit padding)
print(line_list)
````

````python
for i, line in enumerate(line_list):
    line = line[:-4] #remove the last 4 digits (padding)
    num_list = [] 
    for j in range(0, len(line), 6): # step is the 6bit block lenght
        block = line[j:j+6] # then returns a number associated to a number character
        if (len(block) < 6):
            break
        else:
            num_list.append(chr(int(block, 2)))
    num_list = ''.join(num_list)
    num_list = num_list.replace(' ', '-')
    print(f"Credit card number {i} := {num_list}")
````

5. Load the file "user_data.json", filter the data by the "CreditCardType" field equals to "American Express". Than save the data a to CSV.

````python
import json

data = json.load(open('user_data.json'))
data
````

````python
AE = []
for d in data:
    if (d['CreditCardType'] == 'American Express'):
        AE.append(d)

# touch American_Express.csv

with open('American_Express.csv', mode='w') as f:
    f.write(str(data[0].keys()) + " \n ")
    for d in AE:
        f.write(f"{d['ID']},{d['JobTitle']},{d['EmailAddress']},{d['FirstNameLastName']},{d['CreditCard']},{d['CreditCardType']} \n")
````
