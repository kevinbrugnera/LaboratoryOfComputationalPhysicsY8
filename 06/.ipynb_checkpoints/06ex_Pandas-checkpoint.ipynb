## Pandas analysis

This exercise consists in analyzing a dataset containg timing information from a series of Time-to-Digital-Converters (TDC) implemented in a couple of FPGAs. Each measurement (i.e. each row of the input file) consists of a flag that specifies the type of message ('HEAD', which in this case is always 1), two addresses of the TDC providing the signal ('FPGA' and 'TDC_CHANNEL'), and the timing information ('ORBIT_CNT', 'BX_COUNTER', and 'TDC_MEAS'). Each TDC count corresponds to 25/30 ns, whereas a unit of BX_COUNTER corresponds to 25 ns, and the ORBIT_CNT is increased every 'x' BX_COUNTER. This allows to store the time in a similar way to hours, minutes and seconds.

```python
# If you didn't download it yet, please get the relevant file now!
#!wget https://www.dropbox.com/s/xvjzaxzz3ysphme/data_000637.txt
```

```python
import pandas as pd
import numpy as np

data = pd.read_csv('data_000637.txt')
data.head()
```

1\. Create a Pandas DataFrame reading N rows of the 'data_000637.txt' dataset. Choose N to be smaller than or equal to the maximum number of rows and larger that 10k.


```python
#N = 100_000
#df = data[:N] # -> first 10000 rows
df = data # we select the entire dataset
```

2\. Find out the number of BX in a ORBIT (the value 'x').

```python
df.hist('BX_COUNTER') # uniformly distributed since not continous
```

```python
# we will estimate the number of count per orbit viewing the highest/lowest value measured on te clock (like 0s and 60s)
x = df['BX_COUNTER'].max() - df['BX_COUNTER'].min()
print(f"Estimated bx_count in an orbit:= {x}")
```

3\. Find out how much the data taking lasted. You can either make an estimate based on the fraction of the measurements (rows) you read, or perform this check precisely by reading out the whole dataset.

```python
# if we estimate it with fraction, we use the lastone of the subset and then multiply for n (number of subset)

def time(orbit, bx, tdc): # in nanoseconds
    t_s = (orbit * x * 25 + bx * 25 + tdc * 25 / 30) * 10**-9
    print(int(t_s), "seconds")
    gg = t_s // (24*3600)
    hr = (t_s - gg*24*3600) // (3600)
    mn = (t_s - gg*24*3600 - hr*3600) // (60)
    sec = t_s - gg*24*3600 - hr*3600 - mn*60
    print(f"Measure lasted: {gg} days {hr} hr {mn} min {sec} sec")

#we calculate it in the last element (most recent update of the clock)
time(df['ORBIT_CNT'].iloc[-1], df['BX_COUNTER'].iloc[-1], df['TDC_MEAS'].iloc[-1])
```

4\. Create a new column with the absolute time in ns (as a combination of the other three columns with timing information).

```python
df['ABS_TIME [ns]'] = df['ORBIT_CNT'] * x * 25 + df['BX_COUNTER'] * 25 + df['TDC_MEAS'] * 25 / 30
df['ABS_TIME [ns]'] = df['ABS_TIME [ns]'].astype('int64') # to remove scientific notation
df.head()
```

5\. Replace the values (all 1) of the HEAD column randomly with 0 or 1.

```python
df['HEAD'] = np.random.randint(0,2, size=df.shape[0]) # 0,2 -> only 0 and 1
df.head()
```

6\. Create a new DataFrame that contains only the rows with HEAD=1.

```python
df1, df0 = df[df['HEAD'] == 1], df[df['HEAD'] == 0] # we separe in two slices
df1.head()
```

7\. Make two occupancy plots (one for each FPGA), i.e. plot the number of counts per TDC channel

```python
FPGA0, FPGA1 = df[df['FPGA'] == 0], df[df['FPGA'] == 1]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 6))

plt.subplot(121)
plt.title('Occupancy FPGA = 0')
plt.hist(FPGA0['TDC_CHANNEL'], bins=FPGA0['TDC_CHANNEL'].max(), color='y', edgecolor='black')
plt.xlabel('TDC CHANNEL number')
plt.ylabel('TDC CHANNEL frequence')

plt.subplot(122)
plt.title('Occupancy FPGA = 1')
plt.hist(FPGA1['TDC_CHANNEL'], bins=FPGA1['TDC_CHANNEL'].max(), color='orange', edgecolor='black')
plt.xlabel('TDC CHANNEL number')
plt.ylabel('TDC CHANNEL frequence')


plt.tight_layout()
plt.show()
```

8\. Use the groupby method to find out the noisy channels, i.e. the TDC channels with most counts (say the top 3)

```python
df_grouped = df.groupby('TDC_CHANNEL').size() # obtain the time tdc used count

print(f"Noisy channels (Top 3): \n \n {df_grouped.sort_values(ascending=False).head(3)}")
```
