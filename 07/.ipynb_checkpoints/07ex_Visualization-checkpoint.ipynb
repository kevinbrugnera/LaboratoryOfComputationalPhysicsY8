1\. **Kernel Density Estimate**

Produce a KDE for a given distribution (by hand, not using seaborn!):

* Fill a numpy array, x,  of len(N) (with N=O(100)) with a variable normally distributed, with a given mean a standard deviation
* Fill an histogram in pyplot taking properly care about the aesthetic
   * use a meaningful number of bins
   * set a proper y axis label
   * set proper value of y axis major ticks labels (e.g. you want to display only integer labels)
   * display the histograms as data points with errors (the error being the poisson uncertainty)

````python
N = 200
x = np.random.normal(loc=2, scale=3, size=N) 
````

````python
fig, ax = plt.subplots(figsize=(8, 8))
# hist
counts, bins, *_ = ax.hist(x, bins=int(np.sqrt(N)), label='N(2, 3)')
# y label
ax.set_ylabel('Counts')

ax.xaxis.set_major_locator(MultipleLocator(2.000))
ax.xaxis.set_minor_locator(AutoMinorLocator(2))
ax.yaxis.set_major_locator(MultipleLocator(5.000))
ax.yaxis.set_minor_locator(AutoMinorLocator(5))

x_mid = [(bins[i+1] + bins[i]) / 2 for i in range(bins.size - 1)]
ax.errorbar(x_mid, counts, np.sqrt(counts), capsize=5, linestyle='', elinewidth=2) #x, y, yerr no lines
ax.grid(linewidth=0.5)
ax.legend()
````

* for every element of x, create a gaussian with the mean corresponding the element value and std as a parameter that can be tuned. The std default value should be:
$$ 1.06 * x.std() * x.size ^{-\frac{1}{5.}} $$
you can use the scipy function `stats.norm()` for that.
* In a separate plot (to be placed beside the original histogram), plot all the gaussian functions so obtained
* Sum (with np.sum()) all the gaussian functions and normalize the result such that the integral matches the integral of the original histogram. For that you could use the `scipy.integrate.trapz()` method

````python
gaussians = []
std = 1.06 * x.std() * pow(x.size, -0.2)
xs = np.linspace(np.min(x) - 3*std, np.max(x) + 3*std, 500)
for loc in x:
    gaussians.append(stats.norm.pdf(xs, loc=loc, scale=std))
````

````python
fig, ax = plt.subplots(1, 2, figsize=(16, 8))

# HISTOGRAM
ax[0].set_title('Histogram N(2, 3)')
counts, bins, *_ = ax[0].hist(x, bins=int(np.sqrt(N)), label='N(2, 3)')
# y label
ax[0].set_ylabel('Counts')

ax[0].xaxis.set_major_locator(MultipleLocator(2.000))
ax[0].xaxis.set_minor_locator(AutoMinorLocator(2))
ax[0].yaxis.set_major_locator(MultipleLocator(5.000))
ax[0].yaxis.set_minor_locator(AutoMinorLocator(5))

bin_width = np.abs(bins[0] - bins[1])
x_mid = [bins[i] + bin_width / 2 for i in range(bins.size - 1)]
ax[0].errorbar(x_mid, counts, np.sqrt(counts), capsize=5, linestyle='', elinewidth=2) #x, y, yerr no lines
ax[0].grid(linewidth=0.5)
ax[0].legend()

#GAUSSIANS

ax[1].set_title('Gaussians centerd in x random variable')
for f in gaussians:
    ax[1].plot(xs, f)
ax[1].set_ylabel('Density')
ax[1].set_xlabel('x')
ax[1].xaxis.set_major_locator(MultipleLocator(2.000))
ax[1].xaxis.set_minor_locator(AutoMinorLocator(2))
ax[1].yaxis.set_major_locator(MultipleLocator(0.05000))
ax[1].yaxis.set_minor_locator(AutoMinorLocator(5))
ax[1].set_ylim([0, 0.38])

#KDE 

gauss = np.sum(gaussians, axis=0)
I = np.trapezoid(gauss, xs) # y integrated in x
I_hist = bin_width*N
gauss_norm = (gauss / I) * I_hist # since we have not normalized the hist, we have to resize le kde to the one not normalized

print(f"Histogram integral value:= {I_hist}")
print(f"KDE integral value:= {bin_width * I}")

ax[0].plot(xs, gauss_norm, color = 'red', label='KDE gaussian')
````

Produce a scatter plot out of a dataset with two categories

* Write a function that generate a 2D datasets of 2 categories. Each category should distribute as a 2D gaussian with a given mean and std (clearly it is better to have different values means..)
* Display the dataset in a scatter plot marking the two categories with different marker colors.

An example is given below

````python
feature1 = np.random.multivariate_normal([2, 2], [[1, 0], [0, 1]], size=100)
feature2 = np.random.multivariate_normal([0, 0], [[2, 1], [1, 3]], size=100)

plt.figure(figsize=(5,5))
plt.scatter(feature1[:,0], feature1[:,1], color = 'r', alpha=.5, label='feature 1')
plt.scatter(feature2[:,0], feature2[:,1], color = 'g', alpha=.5, label='feature 2')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.tight_layout()
plt.show()
````

````python
N = 6
plt.figure(figsize=(7,7))

for n in range(N):
    mu = np.random.randint(0, 100, size=2)
    A = np.random.randint(0, 100, size=(2, 2))
    sigma = np.dot(A, A.T)
    feature = np.random.multivariate_normal(mu, sigma, size=100)   
    color = ['y', 'g', 'r', 'b', 'k', 'orange']
    plt.scatter(feature[:,0], feature[:,1], c = color[n], alpha=.5, label=f"feature {n}")
    
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.tight_layout()
plt.show()
````

3\. **Profile plot**

Produce a profile plot from a scatter plot.
* Download the following dataset and load it as a pandas dataframe:
```bash
wget https://www.dropbox.com/s/hgnvyj9abatk8g6/residuals_261.npy
```
Note that you should use the `np.load()` function to load the file as a numpy array, call the `.item()` method, and then pass it to the `pd.DataFrame()` constructor.
* Inspect the dataset, you'll find two variables (features)
* Clean the sample by selecting the entries (rows) with the variable "residual" in absolute value smaller than 2
* perform a linear regression of "residuals" versus "distances" using `scipy.stats.linregress()` 

````python
!wget https://www.dropbox.com/s/hgnvyj9abatk8g6/residuals_261.npy
````

````python
df = pd.DataFrame(np.load('residuals_261.npy', allow_pickle=True).item())
df.head(10)
````

````python
df = df[np.abs(df['residuals']) < 2 ]

plt.figure(figsize=(10,6))
plt.scatter(df['distances'], df['residuals'], color = 'r', alpha=.5)
plt.xlabel('distances')
plt.ylabel('residuals')

# Linear regression 
m, q, *_ = stats.linregress(df['distances'], df['residuals'])

plt.plot(df['distances'], m * df['distances'] + q, color='k', label = f'y = {m:.4f} * x {q:.4f}')
plt.legend()
plt.tight_layout()
plt.show()
````

* plot a seaborn jointplot of  "residuals" versus "distances", having seaborn performing a linear regression. The result of the regression should be displayed on the plot

````python
sns.jointplot(x="distances", y="residuals", data=df, kind = 'reg',  color='r', line_kws={"color": "black", "linewidth": 2});
````

* Fill 3 numpy arrays
  * x, serving as an array of bin centers for the "distance" variable. It should range from 0 to 20 with reasonable number of steps (bins)
  * y, the mean values of the "residuals", estimated in slices (bins) of "distance"
  * erry, the standard deviation of the  of the "residuals", estimated in slices (bins) of "distance"
* Plot the profile plot on top of the scatter plot

````python
bins = np.linspace(0, 20, 21)
x = bins[:-1] + 0.5

res = df['residuals']
y = []
for i in range(1, x.size + 1): # digitize from 1 to 20
    dig = np.digitize(df['distances'], bins)
    mask = (dig == i)
    y.append(df['residuals'][mask].mean())

y = np.array(y)

erry = []
for i in range(1, x.size + 1): # digitize from 1 to 20
    dig = np.digitize(df['distances'], bins)
    mask = (dig == i)
    erry.append(df['residuals'][mask].std())

erry = np.array(erry)
````

````python
df = df[np.abs(df['residuals']) < 2 ]

plt.figure(figsize=(10,6))
plt.scatter(df['distances'], df['residuals'], color = 'r', alpha=.5)
plt.xlabel('distances')
plt.ylabel('residuals')

# Linear regression 
m, q, *_ = stats.linregress(df['distances'], df['residuals'])

plt.plot(np.linspace(0, 20, 100), m * np.linspace(0, 20, 100) + q, color='k', label = f'y = {m:.4f} * x {q:.4f}')
plt.errorbar(x, y, erry, color='k', linewidth=1, capsize=2, linestyle='')
plt.legend()
plt.tight_layout()
plt.show()
````
