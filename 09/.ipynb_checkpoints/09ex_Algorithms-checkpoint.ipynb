1\. **Maximum wind speed prediction at the Sprogø station**

The exercise goal is to predict the maximum wind speed occurring every 50 years even if no measure exists for such a period. The available data are only measured over 21 years at the Sprogø meteorological station located in Denmark. 

The annual maxima are supposed to fit a normal probability density function. However such function is not going to be estimated because it gives a probability from a wind speed maxima. Finding the maximum wind speed occurring every 50 years requires the opposite approach, the result needs to be found from a defined probability. That is the quantile function role and the exercise goal will be to find it. In the current model, it is supposed that the maximum wind speed occurring every 50 years is defined as the upper 2% quantile.

By definition, the quantile function is the inverse of the cumulative distribution function. The latter describes the probability distribution of an annual maxima. In the exercise, the cumulative probability $p_i$ for a given year i is defined as $p_i = i/(N+1)$ with $N = 21$, the number of measured years. Thus it will be possible to calculate the cumulative probability of every measured wind speed maxima. From those experimental points, the scipy.interpolate module will be very useful for fitting the quantile function. Finally the 50 years maxima is going to be evaluated from the cumulative probability of the 2% quantile.

Practically, load the dataset:

```python
import numpy as np
max_speeds = np.load('max-speeds.npy')
years_nb = max_speeds.shape[0]
```

Compute then the cumulative probability $p_i$ (`cprob`) and sort the maximum speeds from the data. Use then the  UnivariateSpline from scipy.interpolate to define a quantile function and thus estimate the probabilities.

In the current model, the maximum wind speed occurring every 50 years is defined as the upper 2% quantile. As a result, the cumulative probability value will be:

```python
fifty_prob = 1. - 0.02
```

So the storm wind speed occurring every 50 years can be guessed as:

``` python
fifty_wind = quantile_func(fifty_prob)
```



````python
max_speeds = np.load('max-speeds.npy')
years_nb = max_speeds.shape[0] # 21
````

````python
cprob = np.array([i / (years_nb + 1) for i in range(1, years_nb + 1)])
max_speeds = np.sort(max_speeds) #in order

quantile_func = interpolate.UnivariateSpline(cprob, max_speeds)
fifty_prob = 1. - 0.02
fifty_wind = quantile_func(fifty_prob)

print(f"Maximum wind speed occurring every 50 years:= {fifty_wind:.2f} m/s")
````

````python
speed = np.linspace(np.min(max_speeds), np.max(max_speeds), 100)
plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.scatter(cprob, max_speeds, label='Cumulative Probability')
plt.plot(np.arange(0,1,0.01), quantile_func(np.arange(0,1,0.01)), color='red', label='Quantile function')
plt.scatter(0.98, fifty_wind, marker='x', color='k', label='Fifty year maxima')
plt.ylabel('Max speeds [m/s]')
plt.xlabel('Cumulative Probability')
plt.legend(loc='best')
plt.grid()

plt.subplot(122)
plt.hist(max_speeds, label='speed maxima')
plt.axvline(fifty_wind, color='k', label='Fifty year maxima')
plt.xlabel('speed [m/s]')
plt.legend(loc='best')

plt.show()
````

2\. **Curve fitting of temperature in Alaska** 

The temperature extremes in Alaska for each month, starting in January, are given by (in degrees Celcius):

max:  17,  19,  21,  28,  33,  38, 37,  37,  31,  23,  19,  18

min: -62, -59, -56, -46, -32, -18, -9, -13, -25, -46, -52, -58

* Plot these temperature extremes.
* Define a function that can describe min and max temperatures. 
* Fit this function to the data with scipy.optimize.curve_fit().
* Plot the result. Is the fit reasonable? If not, why?
* Is the time offset for min and max temperatures the same within the fit accuracy?

````python
Tmax = np.array([17, 19, 21, 28, 33, 38, 37, 37, 31, 23, 19, 18])
Tmin = np.array([-62, -59, -56, -46, -32, -18, -9, -13, -25, -46, -52, -58])

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']

plt.figure(figsize=(7, 4))
plt.plot(months, Tmax, label='Maxima', color = 'r')
plt.plot(months, Tmin, label='Minima')
plt.legend()
plt.grid()
plt.show()
````

````python
pi = np.pi

def model(x, a, b, c, d): 
    return a * np.sin(b * x + c) + d

max_params, *_ = optimize.curve_fit(model, np.arange(0, 12, 1), Tmax, p0=(15, pi/12, 0, 30))
min_params, *_ = optimize.curve_fit(model, np.arange(0, 12, 1), Tmin, p0=(50, pi/12, 0, -30))

plt.figure(figsize=(20, 6))

plt.subplot(121)
plt.scatter(months, Tmax, label='Maxima', color = 'r')
plt.plot(np.arange(0, 12, 0.01), model(np.arange(0, 12, 0.01), *max_params), label='Maxima Fit')
plt.legend()
plt.grid()

plt.subplot(122)
plt.scatter(months, Tmin, label='Minima', color = 'r')
plt.plot(np.arange(0, 12, 0.01), model(np.arange(0, 12, 0.01), *min_params), label='Minima Fit')
plt.legend()
plt.grid()
plt.show()
````

3\. **2D minimization of a six-hump camelback function**

$$
f(x,y) = \left(4-2.1x^2+\frac{x^4}{3} \right) x^2 +xy + (4y^2 -4)y^2
$$

has multiple global and local minima. Find the global minima of this function.

Hints:

* Variables can be restricted to $-2 < x < 2$ and $-1 < y < 1$.
* Use numpy.meshgrid() and pylab.imshow() to find visually the regions.
* Use scipy.optimize.minimize(), optionally trying out several of its methods.

How many global minima are there, and what is the function value at those points? What happens for an initial guess of $(x, y) = (0, 0)$ ?


````python
def f(v):
    x, y = v
    return (4 - 2.1 * x**2 + x**4 / 3) * x**2 + x*y + (4 * y**2 - 4) * y**2

x, y = np.linspace(-2,2, 1000), np.linspace(-1,1, 1000)
X, Y = np.meshgrid(x, y)

plt.figure(figsize=(10, 5))
plt.imshow(f((X,Y)), extent=[-2, 2, -1, 1])
plt.colorbar(label='f(x,y)')
plt.grid()
plt.xlabel('x')
plt.xlabel('xy')
plt.xlim([-2,2])
plt.ylim([-1,1])
plt.show()
````

````python
methods = ['Nelder-Mead', 'Powell']
x0 = [(0, -0.75), (-1.5, -0.75), (0, 0.75), (1.5, 0.75), (0,0)]

for x in x0:
    res = optimize.minimize(f, x, method=methods[0])
    print(res)

for x in x0:
    res = optimize.minimize(f, x, method=methods[1])
    print(res)
````

4\. **FFT of a simple dataset**

Performe a periodicity analysis on the lynxs-hares population

````python
from scipy.fftpack import fft, ifft, fftfreq, fft2, ifft2
````

````python
data = np.loadtxt('populations.txt')
plt.figure(figsize=(6,4))
plt.plot(data[:,0], data[:,1], label='hare')
plt.plot(data[:,0], data[:,2], label='lynx')
#plt.plot(data[:,0], data[:,3], label='carrot')
plt.xlabel('years')
plt.ylabel('population')
plt.legend()
plt.show()
````

````python
fft_h, fft_l = fft(data[:,1]), fft(data[:,2])
freq = fftfreq(data.shape[0], d=1) # number / year

freq_pos = freq[freq > 0]
ampl_h, ampl_l = np.abs(fft_h[:len(freq_pos)]), np.abs(fft_l[:len(freq_pos)]) #first half are amplitidues of positive freq

peak_freq_h, peak_freq_l = freq_pos[np.argmax(ampl_h)], freq_pos[np.argmax(ampl_l)]

print(f"Dominant hares freq := {peak_freq_h} \n relative period: {1 / peak_freq_h} yr")
print(f"Dominant lynxes freq := {peak_freq_l} \n relative period: {1 / peak_freq_l} yr")
````

````python
plt.figure(figsize=(6,4))
plt.plot(freq_pos, ampl_h, marker='.', color='green', label='Hares')
plt.plot(freq_pos, ampl_l, marker='.', color='red', label='Lynxs')

plt.xlabel('Freq [1/yr]')
plt.ylabel('Amplitude')
plt.suptitle('Lynxs vs Hares Period Analysis')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````

5\. **FFT of an image**

* Examine the provided image `moonlanding.png`, which is heavily contaminated with periodic noise. In this exercise, we aim to clean up the noise using the Fast Fourier Transform.
* Load the image using pylab.imread().
* Find and use the 2-D FFT function in scipy.fftpack, and plot the spectrum (Fourier transform of) the image. Do you have any trouble visualising the spectrum? If so, why?
* The spectrum consists of high and low frequency components. The noise is contained in the high-frequency part of the spectrum, so set some of those components to zero (use array slicing).
* Apply the inverse Fourier transform to see the resulting image.

````python
figure = pylab.imread('moonlanding.png', format="PNG")
pylab.imshow(figure, cmap='grey') 
figure.shape
````

````python
sig = fft2(figure)
# freq_x, freq_y = fftfreq(sig.shape[0]), fftfreq(sig.shape[1])
ampl = np.abs(sig)

plt.figure(figsize=(20,6))
plt.subplot(121)
plt.imshow(ampl)  # we cannot se anything
plt.title("2D Power Spectrum")
plt.colorbar(label='Amplitude')
plt.subplot(122)
plt.imshow(np.log( 1 + ampl ))  # use log scale to make the graph readable
plt.title("2D Power Spectrum (Log Scale)")
plt.colorbar(label='Amplitude')
plt.show()
````

````python
threshold = 0.01
sig[ampl > np.max(ampl) * threshold] = 0

plt.imshow(np.log(1 + np.abs(sig)))  # we cannot se anything
plt.title("2D Power Spectrum")
plt.colorbar(label='Filtered Amplitude')
````

````python
filt_figure = ifft2(sig).real
pylab.imshow(filt_figure, cmap='grey')
````
